```python
orders = [
    ("Wireless Mouse", 2, 799),
    ("Basmati Rice 5kg", 8, 650),
    ("Denim Jeans", 7, 1599),
    ("Bluetooth Speaker", 5, 2499),
    ("Notebook Pack", 10, 120),
    ("USB-C Cable", 0, 299)
]

total = 0

for i, (name, qty, price) in enumerate(orders, 1):
    if qty <= 0 or price <= 0:
        continue

    sub = qty * price

    if sub >= 10000:
        discount = sub * 0.10
    elif sub >= 5000:
        discount = sub * 0.05
    else:
        discount = 0

    gst = (sub - discount) * 0.18
    final = sub - discount + gst
    total += final

    print(name, sub, discount, gst, round(final, 2))

print("Grand Total:", round(total, 2))
```
